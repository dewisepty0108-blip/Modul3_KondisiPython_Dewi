```python
# Kondisi if
nilai = 9

# Jika kondisi benar, perintah akan dijalankan
if nilai > 7:
    print("Sembilan Lebih Besar Dari Angka Tujuh")

# Jika kondisi salah, perintah tidak dijalankan
if nilai > 10:
    print("Sembilan Lebih Besar Dari Angka Sepuluh")
```

```python
# Kondisi if else
nilai = 3

if nilai > 7:
    print("Selamat Anda Lulus")
else:
    print("Maaf Anda Tidak Lulus")
```
